### Data limitation Every user id and product id appears exactly once, so per-product and per-user analysis is not possible. The analysis therefore focuses on interaction types and time trends.

```python
print("Total rows        :", len(df))
print("Unique user ids   :", df["user id"].nunique())
print("Unique product ids:", df["product id"].nunique())

user_interactions = df["user id"].value_counts()
print("\nInteractions per user:")
print(user_interactions.describe())
```

Visualisasi

```python
import matplotlib.pyplot as plt

interaction_counts = df['Interaction type'].value_counts()

interaction_counts.plot(kind='bar')

plt.title('Distribution of User Interactions')
plt.xlabel('Interaction Type')
plt.ylabel('Number of Interactions')
plt.xticks(rotation=0)
plt.show()
```

```python
daily_interactions = (
    df.groupby(df['Time stamp'].dt.date)
    .size()
)

plt.figure(figsize=(12, 5))
daily_interactions.plot()

plt.title('Daily User Interactions')
plt.xlabel('Date')
plt.ylabel('Number of Interactions')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
monthly_interactions = (
    df.groupby(df['Time stamp'].dt.to_period('M'))
    .size()
)
monthly_interactions.index = monthly_interactions.index.astype(str)

plt.figure(figsize=(10, 5))
monthly_interactions.plot(kind="bar")

plt.title('Monthly User Interactions')
plt.xlabel('Month')
plt.ylabel('Number of Interactions')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
interaction_counts = df['Interaction type'].value_counts()

purchase_rate = (
    interaction_counts['purchase'] / interaction_counts.sum()
) * 100

purchase_rate
```

```python
monthly_purchase = (
    df[df['Interaction type'] == 'purchase']
    .groupby(df['Time stamp'].dt.to_period('M'))
    .size()
)
monthly_purchase.index = monthly_purchase.index.astype(str)

plt.figure(figsize=(10, 5))
monthly_purchase.plot(kind="bar")
plt.title('Monthly Purchases')
plt.xlabel('Month')
plt.ylabel('Number of Purchases')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
monthly_interaction_type = (
    df.groupby([
        df['Time stamp'].dt.to_period('M'),
        'Interaction type'
    ])
    .size()
    .unstack(fill_value=0)
)
monthly_interaction_type.index = monthly_interaction_type.index.astype(str)

monthly_interaction_type.plot(kind="line", figsize=(12, 5), marker="o")
plt.title("Monthly User Interaction Trends")
plt.xlabel("Month")
plt.ylabel("Number of Interactions")
plt.legend(title="Interaction Type")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
total = len(df)
q4_mask = (df["Time stamp"] >= "2023-10-01") & (df["Time stamp"] < "2024-01-01")
q4_share = q4_mask.sum() / total * 100
peak_month = monthly_interactions.idxmax()

monthly_purchase_rate = monthly_interaction_type["purchase"] / monthly_interaction_type.sum(axis=1) * 100
top3 = monthly_interactions.nlargest(3).index

print(f"Overall purchase rate         : {purchase_rate:.1f}%")
print(f"Busiest month                 : {peak_month} ({monthly_interactions.max()} interactions)")
print(f"Q4 2023 share of interactions : {q4_share:.1f}% ({q4_mask.sum()} of {total})")
print("Purchase rate in 3 busiest months:")
print(monthly_purchase_rate.loc[top3].round(1).to_string())
```

**Takeaways**

- Interactions peak in Q4 2023, with November the busiest month.
- The purchase rate stays near 30% even in peak months, so the surge comes from more traffic, not higher conversion.
- January 2023 is also high-volume, so more than one year of data is needed to confirm seasonality.
