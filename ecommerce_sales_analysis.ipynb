Cleaning data

```python
import pandas as pd
```

```python
df = pd.read_csv("E-commerece sales data 2024.csv")

df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df[df.isnull().any(axis=1)].head(20)
```

```python
df[df["Interaction type"].isnull()].head(20)
```

```python
df[df["user id"].isnull() & df["product id"].isnull() & df["Time stamp"].isnull()].head(20)
```

```python
df.isna().all(axis=1).sum()
```

```python
df = df.dropna(how="all")
```

```python
df.shape
```

```python
df = df.drop(columns=["Unnamed: 4"])
```

```python
df.shape
```

```python
df["Interaction type"].value_counts(dropna=False)
```

```python
df[df["Interaction type"].isnull()].head(20)
```

```python
df = df.dropna(subset=["Interaction type"])
```

```python
df.shape
```

```python
df.duplicated().sum()
```

```python
df.dtypes
```

```python
df["Time stamp"] = pd.to_datetime(
    df["Time stamp"],
    dayfirst=True,
    errors="coerce"
)
```

```python
df.dtypes
```

```python
df.isnull().sum()
```

Visualisasi

```python
import matplotlib.pyplot as plt

interaction_counts = df['Interaction type'].value_counts()

interaction_counts.plot(kind='bar')

plt.title('Distribution of User Interactions')
plt.xlabel('Interaction Type')
plt.ylabel('Number of Interactions')
plt.xticks(rotation=0)
plt.show()
```

```python
daily_interactions = (
    df.groupby(df['Time stamp'].dt.date)
    .size()
)

plt.figure(figsize=(12, 5))
daily_interactions.plot()

plt.title('Daily User Interactions')
plt.xlabel('Date')
plt.ylabel('Number of Interactions')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
montly_interactions = (
    df.groupby(df['Time stamp'].dt.to_period('M'))
    .size()
)

plt.figure(figsize=(10, 5))
montly_interactions.plot(kind="bar")

plt.title('Montly User Interactions')
plt.xlabel('Month')
plt.ylabel('Number of Interactions')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
interaction_counts = df['Interaction type'].value_counts()

purchase_rate = (
    interaction_counts['purchase'] / interaction_counts.sum()
) * 100

purchase_rate
```

```python
monthly_purchase = (
    df[df['Interaction type'] == 'purchase']
    .groupby(df['Time stamp'].dt.to_period('M'))
    .size()
)

plt.figure(figsize=(10, 5))
monthly_purchase.plot(kind="bar")

plt.title('Monthly Purchase Interactions')
plt.xlabel('Month')
plt.ylabel('Number of Interactions')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
top_products = (
    df["product id"]
    .value_counts()
    .head(10)
)

plt.figure(figsize=(10, 5))
top_products.sort_values().plot(kind="barh")

plt.title('Top 10 Most Interacted Products')
plt.xlabel('Number of Interactions')
plt.ylabel('Product ID')
plt.tight_layout()
plt.show()
```

```python
monthly_interaction_type = (
    df.groupby([
        df['Time stamp'].dt.to_period('M'),
        'Interaction type'
    ])
    .size()
    .unstack(fill_value=0)
)

monthly_interaction_type.plot(
    kind="line",
    figsize=(12, 5),
    marker="o"
)

plt.title("Monthly User Interactions Trends")
plt.xlabel("Month")
plt.ylabel("Number of Interactions")
plt.legend(title="Interaction Type")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
